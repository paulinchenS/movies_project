# Data Exploration Findings

**Dataset was downloaded** October 8, 2026. Findings are from that time. 

There are 1511625 (over 1.5 million) entries with 24 columns. The column names are as follows: 'id', 'title', 'vote_average', 'vote_count', 'status', 'release_date', 'revenue', 'runtime', 'adult', 'backdrop_path', 'budget', 'homepage', 'imdb_id', 'original_language', 'original_title', 'overview', 'popularity', 'poster_path', 'tagline', 'genres', 'production_companies', 'production_countries', 'spoken_languages', 'keywords'.

## Duplicate entries
There are 384 duplicate rows. There are 1376 rows with repeated IDs, excluding the first occurrence of each ID, and 231780 repeated titles, excluding the first occurrence of each title. The ids are worth investigating, but duplicate titles are nothing unusual. When counting duplicate ids, we saw that there are at most 4 entries with the same id. An example is id=1236071 which shows four different entries for the movie "Semetua". The first 2 seem to be identical, the other 2 do have the same title but are mostly empty for other columns. These problems need to be tackled during cleaning.

## Missing values
Many movies are missing homepage, tagline, keywords, backdrop_path, production_companies, imdb_id, production_countries and spoken_languages. Importance of these columns depends on analysis. A more interesting column is genres, which is missing for 680469 movies. poster_path, release_date and overview are also missing for a lot of movies. title and original_title are missing for 20 entries. 1 movie has negative revenue, 1486059 movies have revenue 0. There are only 18509 movies that have both positive revenue and budget. 1 movie has negative runtime, 480695 have runtime 0. Only 361176 movies have at least one vote.
