```python
title1 = "La Veganza De Ebaado"
title2 = "La Venganza de Ebaado"

similarity = Levenshtein.ratio(
    title1.casefold(),
    title2.casefold()
)

print(similarity)
```

```python
def jaccard_similarity(title1, title2):
    words1 = set(title1.casefold().split())
    words2 = set(title2.casefold().split())

    if not words1 and not words2:
        return 1.0

    return len(words1 & words2) / len(words1 | words2)
```

```python
jaccard_similarity(title1, title2)
```

```python
from itertools import combinations
```

```python
group_sizes = imdb_duplicates.groupby('imdb_id').size()
large_ids = group_sizes[group_sizes >= 3].index
example = imdb_duplicates[imdb_duplicates['imdb_id'] == large_ids[0]]
print(example[['id', 'original_title', 'release_date']])
pairs = list(combinations(example.index, 2))
print('Records:', len(example))
print('Possible pairs:', len(pairs))
print('First five pairs:', pairs[:5])
for idx1, idx2 in pairs:
    title1 = example.loc[idx1, 'original_title']
    title2 = example.loc[idx2, 'original_title']
    similarity = Levenshtein.ratio(
        title1.casefold(),
        title2.casefold())
    print(title1, '|', title2, ':', round(similarity, 3))
```

```python
pair_results = []
for imdb_id, group in imdb_duplicates.groupby('imdb_id'):
    for idx1, idx2 in combinations(group.index, 2):
        title1 = group.loc[idx1, 'original_title']
        title2 = group.loc[idx2, 'original_title']
        if pd.isna(title1) or pd.isna(title2):
            continue
        similarity = Levenshtein.ratio(
            title1.casefold(),
            title2.casefold())
        year1 = pd.to_datetime(
            group.loc[idx1, 'release_date'], errors='coerce'
        ).year
        year2 = pd.to_datetime(
            group.loc[idx2, 'release_date'], errors='coerce'
        ).year

        pair_results.append({
            'imdb_id': imdb_id,
            'idx1': idx1,
            'idx2': idx2,
            'similarity': similarity,
            'year1': year1,
            'year2': year2})
pair_df = pd.DataFrame(pair_results)
```

```python
pair_df['year_difference'] = (pair_df['year1'] - pair_df['year2']).abs()
matches = pair_df[
    (pair_df['similarity'] >= 0.80)
    & ((pair_df['year_difference'] <= 1)
        | pair_df['year_difference'].isna())]
```

```python
print('Total pairs compared:', len(pair_df))
print('Potential duplicate pairs:', len(matches))
print('IMDb groups with matches:', matches['imdb_id'].nunique())
```

```python
matching_pairs = {
    frozenset((row.idx1, row.idx2))
    for row in matches.itertuples()
}
```

```python
movie_columns = [
    col for col in df_new.columns
    if col != 'completeness_score']
scores = df_new[movie_columns].notna().sum(axis=1)
rows_to_remove = []
for imdb_id, group in imdb_duplicates.groupby('imdb_id'):
    ranked_indices = scores.loc[group.index].sort_values(
        ascending=False,
        kind='stable'
    ).index
    kept_indices = []
    for idx in ranked_indices:
        is_duplicate = any(
            frozenset((idx, kept_idx)) in matching_pairs
            for kept_idx in kept_indices)
        if is_duplicate:
            rows_to_remove.append(idx)
        else:
            kept_indices.append(idx)
```

```python
print('Rows proposed for removal:', len(rows_to_remove))
print('Rows currently in df_new:', len(df_new))
print('Rows after cleaning:', len(df_new) - len(rows_to_remove))
```

# Cleaning rules:

First remove duplicate rows.

## ID
After removing duplicate rows, 989 IDs still appear multiple times. Of the 989 IDs, 147 have different titles. When normalizing the titles, the number reduced to 142. Instead normalizing original_title, reduces the count to 72. Manually checking these titles show that differences come from: spelling errors, formatting errors, alternate naming, etc. We therefore assume that all IDs refer to the same movie (not proven!)

For entries with same ID, we sort by "completeness score" (how many non nans for all columns) and keep the entry with highest score. This removed 992 rows and no duplicate IDs d. 

## IMDB ID
One imdb_id can lead to many id. To compare the movies, we can use Levenshtein similiarity with some treshold like 0.8 and maybe a release year difference (remakes might have the same title but a different release date). Again, entries with a high completeness score are kept. This removes 4331 rows. An example shows that we are not able to catch all duplicates, due to alternative naming that may fall below the 0.8 threshold. Still, we are able to remove quite a large number of duplicates like this.
